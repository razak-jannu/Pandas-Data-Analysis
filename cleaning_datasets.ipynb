7. Handling Missing Data Through functions
   - fillna()
   - dropna()
   - replace()


```python
import pandas as pd
csv_2=pd.read_csv("dirty_sales_data.csv")
```

```python
csv_2.isna().sum()
csv_2.head(40)
```

```python
# Q1. Check the number of missing values in each column.
csv_2.isna().sum()
# Q2. Fill missing values in the Quantity column with 0.
csv_2["Quantity"].fillna(0)

# Q3. Fill missing values in the Price_Per_Item column with 0.
csv_2["Price_Per_Item"].fillna(0)

# Q4. Find the mean of Quantity and fill missing values with the mean.
# qun_mean=csv_2["Quantity"].mean()
# csv_2["Quantity"]=csv_2["Quantity"].fillna((round(qun_mean,0)))
# # print(round(qun_mean,0))

# Q5. Find the median of Price_Per_Item and fill missing values with the median.
csv_2["Price_Per_Item"]=csv_2["Price_Per_Item"].str.replace("$","",regex=False)
csv_2["Price_Per_Item"]=csv_2["Price_Per_Item"].astype(float)
ppi_median=csv_2["Price_Per_Item"].median()
# print(ppi_median)
csv_2["Price_Per_Item"]=csv_2["Price_Per_Item"].fillna(ppi_median)
# Q6. Fill missing values in Product_Category with "Unknown".
csv_2["Product_Category"]=csv_2["Product_Category"].fillna("Unknown")
# Q7. Fill missing values in Customer_Name with "Unknown".
csv_2["Customer_Name"]=csv_2["Customer_Name"].fillna("Unknown")

# Q8. Remove all rows that contain missing values.
csv_2.dropna(inplace=True)

# Q9. Remove rows where Quantity is missing.
csv_2.dropna(subset=["Quantity"],inplace=True)

# Q10. Remove rows where Price_Per_Item is missing.
csv_2.dropna(subset=["Price_Per_Item"],inplace=True)

# Q11. Remove columns that contain missing values.
csv_2.dropna(axis=1,inplace=True)
# Q12. Replace "unknown" with "Unknown".
csv_2.replace("unknown","Unknown",inplace=True)

# Q13. Replace empty/invalid category values with "Unknown".
csv_2.fillna("Unknown",inplace=True)

# Q14. Find any incorrect Product_Category value and replace it with the correct value.
csv_2["Product_Category"]=csv_2["Product_Category"].replace("Unknown","Office Supplies")
# Q15. Create a copy of df and handle missing values without changing the original df.
df_cleaned=csv_2.copy()
# Q16. Fill missing values:
# Quantity → mean
df_cleaned["Quantity"]=df_cleaned["Quantity"].fillna(df_cleaned["Quantity"].mean())
# Price_Per_Item → median
df_cleaned["Price_Per_Item"]=df_cleaned["Price_Per_Item"].fillna(df_cleaned["Price_Per_Item"].median())
# Product_Category → "Unknown"
df_cleaned["Product_Category"]=df_cleaned["Product_Category"].fillna("Unknown")

# Q17. Check again and verify that no missing values are left.
csv_2.isna().sum()
df_cleaned.isna().sum()
```
